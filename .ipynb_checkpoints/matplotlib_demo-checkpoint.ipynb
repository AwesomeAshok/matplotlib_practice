```python
import matplotlib.pyplot as plt

year = [1995,2000, 2005, 2010,2015]
lang = ["None","Java", "c", "c++", "Python"]
color=['g','blue','r','pink']
plt.bar(year, lang,color=color,edgecolor="black")#align="edge" to align the line in edge of the center line

plt.xlabel("Year")
plt.ylabel("Language")
plt.title("coding Language comarison")

plt.show()
```

```python
plt.figure()
plt.scatter([10,20,30,40,50,60,70,80],[1,2,2.1,3,3.5,4,6,1.2],color='g',marker="+",s=100) 
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("DEMO GRAPH")
plt.show();
```

```python
x=[1,2,3,4,5]
y=[10,30,20,50,40]
plt.plot(x,y,color='r',marker=".",linestyle="dotted")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Example for plot")
plt.show();
```

```python
country=["India","USA","Japan","China"]
population=[140,60,35,130]
color=['g','blue','violet','pink']
exp=[0.05,0,0,0]
plt.pie(population,labels=country,colors=color,explode=exp)
plt.
plt.show();
```

```python
stockA=[1,4,5,10,7,2]
stockB=[1,10,6,4,3,7]
stockC=[1,3,7,8,9]
plt.title("Stock Market Graph")
plt.plot(stockA,label="stock A")
plt.plot(stockB,label="stock B")
plt.plot(stockC,label="stock C")
plt.legend()
plt.show();
```
